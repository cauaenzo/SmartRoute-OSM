# SmartRoute-OSM: Extração e Persistência de Dados Espaciais

Neste primeiro notebook do pipeline, faremos o download do grafo de vias dirigíveis de **Quixadá, Ceará** via `OSMnx`. Em seguida, enriqueceremos o grafo com métricas de velocidade e tempo de viagem estimados para, finalmente, salvá-lo em disco na pasta `data/`.


```python
import os
import osmnx as ox

# Configurações de cache do OSMnx
ox.settings.use_cache = True
ox.settings.log_console = False

# Criar a pasta 'data' caso não exista
data_dir = "../data"
if not os.path.exists(data_dir):
    os.makedirs(data_dir)
    print(f"Diretório '{data_dir}' criado com sucesso.")
else:
    print(f"Diretório '{data_dir}' já existe.")
```

## 1. Download do Grafo de Vias Urbanas
Baixamos o grafo de vias navegáveis (`drive`) para o município de Quixadá.

```python
# Definir local do estudo
place_name = "Quixadá, Ceará, Brazil"

print(f"Obtendo dados do OpenStreetMap para: {place_name}...")
G = ox.graph_from_place(place_name, network_type="drive")

print(f"Grafo extraído com sucesso! Total de nós: {len(G.nodes)}, Total de arestas: {len(G.edges)}")
```

## 2. Enriquecimento de Atributos (Velocidade e Tempo de Viagem)
Adicionamos estimativas de velocidade média padrão por tipo de via (`speed_kph`) e calculamos o tempo necessário para percorrer cada aresta em segundos (`travel_time`).

```python
# Adiciona velocidades padrão (km/h) baseadas no tipo de via
G = ox.add_edge_speeds(G)

# Calcula o tempo de viagem em segundos para cada aresta
G = ox.add_edge_travel_times(G)

print("Atributos 'speed_kph' e 'travel_time' adicionados às arestas do grafo.")
```

## 3. Persistência dos Dados
Salvamos o grafo enriquecido no formato `.graphml` dentro do diretório `data/`. Esse arquivo servirá como entrada para os notebooks de cálculo de menor caminho.

```python
output_path = os.path.join(data_dir, "quixada_drive.graphml")

# Salvar o grafo em arquivo GraphML
ox.save_graphml(G, filepath=output_path)

print(f"Grafo salvo com sucesso em: {output_path}")
```
